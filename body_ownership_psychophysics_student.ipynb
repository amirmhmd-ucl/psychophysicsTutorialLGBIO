## 1. What is body ownership?

The feeling that your hand, your arm, or your whole body belongs to *you* is so automatic that it rarely enters conscious awareness. This experience is called **body ownership**: the sense that a given body part is part of one's own body, as opposed to an external object.

Body ownership is not a single sense in the way vision or hearing is. Instead, it is thought to emerge from the integration of signals across several sensory modalities, including vision, touch, and proprioception (the sense of limb position and movement). When these signals are consistent with one another — for instance, when you see your hand being touched at exactly the moment you feel the touch, and in the expected spatial location — the brain concludes that the seen limb and the felt limb are one and the same, and ownership follows.

This becomes strikingly clear in the **rubber hand illusion** (Botvinick & Cohen, 1998). If a visible rubber hand (see picture below) is stroked with a paintbrush in perfect synchrony with strokes applied to a participant's own hidden hand, most people quickly begin to feel that the rubber hand is their hand, and that it is the one being touched. The illusion degrades when breaking the synchrony, rotating the rubber hand into an anatomically implausible posture, or moving it too far away.

![Rubber hand illusion setup](https://raw.githubusercontent.com/amirmhmd-ucl/psychophysicsTutorialLGBIO/main/images/rubber_hand_illusion.png) *source: www.bio.psy.ruhr-uni-bochum.de*

Although the broad principle — multisensory integration — is well established, the exact rules governing this integration are not. Two open questions motivate this exercise:

- **What are the precise spatial, temporal, and material constraints** under which sensory signals are bound into a single sense of ownership?
- **Is body ownership itself a genuine perceptual phenomenon** — something that can be measured with the same rigor as brightness or loudness discrimination — or is it a vaguer, higher-level cognitive judgment?

Answering these questions requires a measurement tool that is sensitive, unbiased, and quantitative. As you'll see below, this is precisely where classical methods fall short, and where psychophysics offers a solution.

## 2. Classical approaches to measuring body ownership

Historically, three main types of measures have been used to assess the strength of the rubber hand illusion:

**Questionnaires.** Participants rate statements such as *"It seemed as if I were feeling the touch of the paintbrush in the location where I saw the rubber hand touched"* on a Likert scale, typically after the illusion has been induced. Questionnaires are easy to administer and directly probe subjective experience, but they suffer from several well-known problems: participants have limited introspective access to subtle perceptual changes, ratings are collected retrospectively (relying on memory of a fading experience), and responses are vulnerable to demand characteristics — participants unconsciously giving the answer they think the experimenter wants.

**Proprioceptive drift.** With eyes closed, participants indicate the perceived position of their real (hidden) hand before and after illusion induction. If ownership has transferred to the rubber hand, perceived hand position typically drifts toward it. This gives an objective, behavioral index, but its validity as a *causal* marker of ownership has been questioned: drift can occur even in control conditions where ownership is unlikely to arise, and manipulations that change proprioceptive drift do not always change subjective ownership ratings, and vice versa.

**Skin conductance response (SCR).** A threat directed at the rubber hand (e.g., a needle prick) evokes a measurable autonomic response — but only if the hand is felt to be one's own. SCR is objective and physiological, but it is nonspecific (many emotional states other than ownership-linked threat evoke an SCR), it habituates quickly with repeated trials, and individuals vary enormously in their baseline responsiveness.

Taken together, these methods share a common limitation: none of them directly and repeatedly asks the perceptual question we actually care about — *"does this feel like my hand, right now?"* — in a way that is resistant to bias and can be sampled many times per participant. This is the gap that a psychophysical discrimination task is designed to fill.

## 3. A psychophysical alternative: the two-alternative forced-choice (2AFC) task

Chancel and Ehrsson (2020) reasoned that if body ownership is a genuine perceptual phenomenon, it should be measurable using the same rigorous tools that psychophysics has long used to study vision, hearing, and touch. They designed a **two-alternative forced-choice discrimination task** that allows precise and direct measurement of body ownership, in which participants decide which of two rubber hands feels more like their own.

The setup: a participant's real hand rests hidden under a table, while two identical rubber hands are placed in view, one to the left and one to the right. On each trial, all three hands (the two rubber hands and the real hand) are touched by small robotic probes for several seconds. The critical manipulation is the **timing** of these touches: one rubber hand is touched perfectly in synchrony with the real hand, while the other is touched with a systematically varied delay (e.g., 0, 50, 100, or 200 ms). After the stimulation period, the participant simply reports: *"Which hand feels more like mine — left or right?"*

![Two-alternative forced-choice setup with two rubber hands above the actual hand](https://raw.githubusercontent.com/amirmhmd-ucl/psychophysicsTutorialLGBIO/main/images/2afc_setup.png)

Because the two rubber hands are directly pitted against each other on every trial, this task sidesteps many of the weaknesses of classical measures:

- It yields a simple, forced binary choice rather than a graded subjective rating, which is less susceptible to response bias and interpretation differences across participants.
- It can be repeated many times per participant under systematically varied conditions (asynchrony, distance, tactile material), producing rich, quantitative datasets rather than a single post-hoc score.
- It relates directly to *perception in the moment*, rather than to memory of a past experience or an indirect physiological correlate.

The same logic can be extended to test other congruence principles besides timing — for example, the *spatial* distance between the rubber hands and the real hand, or the *material congruence* between what is seen touching the rubber hand and what is felt touching the real hand. In each case, the manipulated variable plays the role of the physical stimulus dimension in a classical psychophysics experiment, and the participant's binary choices are the perceptual response.

## 4. Psychophysics fundamentals: fitting a psychometric function

Once we have binary choice data collected across a range of stimulus values (e.g., visuotactile asynchrony from $-200$ to $+200$ ms), the goal of psychophysics is to describe *how the probability of a given response changes as a function of the stimulus*. This relationship is captured by the **psychometric function**.

For the 2AFC ownership task, we compute, at each tested asynchrony $x$, the proportion of trials on which the participant chose (say) the right rubber hand. These proportions typically trace out an S-shaped curve, which can be well described by a cumulative Gaussian distribution:

$$
P(x) = \lambda + (1 - 2\lambda) \cdot \Phi\left(\frac{x - \mu}{\sigma}\right)
$$

where $\Phi$ is the standard normal cumulative distribution function, and:

- $\mu$ is the **point of subjective equality (PSE)** — the stimulus value at which the participant is equally likely to choose either hand. Under our sign convention, a nonzero PSE reveals a systematic *bias* (e.g., a general preference for whichever hand is closer to the body midline).
- $\sigma$ is the **discrimination threshold** — how much asynchrony is needed to reliably shift the participant's choice. A smaller $\sigma$ means the participant is more sensitive to the manipulated variable; this is closely related to the classical notion of the just-noticeable difference (JND).
- $\lambda$ is the **lapse rate** — the rate of stimulus-independent errors (inattention, motor slips), which prevents the fit from being distorted by a small number of "impossible" responses at the extremes.

Fitting this function to data means finding the values of $\mu$, $\sigma$, and $\lambda$ that make the observed pattern of choices most likely, typically via maximum-likelihood estimation. In the exercise, you will implement this fitting procedure yourself (for example using `scipy.optimize.curve_fit` or a maximum-likelihood approach with `scipy.stats.norm`), apply it to experimental data, and interpret $\mu$ and $\sigma$ in terms of the perceptual and cognitive mechanisms of body ownership.

## Bonus question: from fixed sampling to adaptive methods

The approach above uses what is called the **method of constant stimuli**: a fixed, predetermined set of stimulus values (e.g., seven asynchronies) is presented repeatedly, in random order, the same number of times to every participant. This method is simple and produces data that are easy to visualize and fit — but it is statistically inefficient. Many trials are "wasted" on stimulus values far from the region that actually matters for estimating $\mu$ and $\sigma$ (i.e., far from the steep part of the psychometric curve), while relatively few trials fall exactly where they would be most informative.

This inefficiency also matters in practice. Psychophysical sessions are long and effortful for participants (in the original study, each experimental block already took 27 minutes), and any method that can extract the same precision from fewer trials directly translates into shorter, less fatiguing, and more scalable experiments.

This motivates **adaptive psychophysics experiments**. Instead of fixing the stimulus set in advance, adaptive methods use the participant's ongoing responses to decide, trial by trial, which stimulus value to present next — concentrating sampling around the region of the psychometric function that is currently most uncertain (typically near the PSE). Classic examples include rule-based procedures such as the transformed up-down staircase (Levitt, 1971) and PEST (Taylor & Creelman, 1967), which change the stimulus according to fixed rules, and Bayesian adaptive methods such as QUEST (Watson & Pelli, 1983), which maintain and update a probabilistic estimate of the psychometric function after every trial.
